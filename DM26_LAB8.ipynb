```python
# Load the College.csv file

import pandas as pd

college = pd.read_csv('/content/College.csv')
college.head()
```

```python
college.columns
```

```python
# Set college names as the DataFrame index
college.set_index('Unnamed: 0', inplace=True)
college.head()
```

```python
# X and y split
y = college['Private']
x = college.drop('Private', axis=1)
```

```python
# 스케일링 하기 전에 데이터 확인
import matplotlib.pyplot as plt

x.boxplot()
plt.show()
```

```python
# Scaling x
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
x_scaled = scaler.fit_transform(x)
```

```python
# 스케일링한 후 데이터 확인
plt.boxplot(x_scaled)
plt.show()
```

```python
# Encoding y
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
y = encoder.fit_transform(y)
y
```

```python
encoder.classes_ #0 공립 1 사립
```

```python
# Train and test split
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.2, random_state=10)
```

```python
# 데이터가 잘 나눠졌는지 확인
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
import seaborn as sns

sns.countplot(x=y_train)
plt.show()
```

```python
# Resample imbalanced data (SMOTE)
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=10)
x_smote, y_smote = smote.fit_resample(x_train, y_train)
```

```python
sns.countplot(x=y_smote)
plt.show()
```

```python
# Build a NN Model (First layer: 10 nodes, with relu activation, 2nd layer: 10 nodes, with relu activation, 3rd layer: output, with sigmoid activation). 신경망모델 구축

from keras.models import Sequential
from keras.layers import Dense

model = Sequential()
model.add(Dense(10, input_dim=x_smote.shape[1], activation='relu'))
model.add(Dense(10, activation='relu'))
model.add(Dense(10, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
model.summary()
```

```python
# 모델 컴파일
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# Fit the model (epochs=100). 모델훈련
history = model.fit(x_smote, y_smote, epochs=100, validation_data=(x_test, y_test))
```

```python
# history graph
import matplotlib.pyplot as plt

plt.subplot(121)
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Test'], loc='upper left')

plt.subplot(122)
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Test'], loc='upper left')

plt.show()
```

```python
# Evaluate the results(accuracy_score, confusion_matrix, classification_report, history graph) 결과평가
```

```python
# accuracy_score
model.evaluate(x_test, y_test)
```

```python
# confusion_matrix
from sklearn.metrics import confusion_matrix

y_pred = model.predict(x_test).round()
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d')
plt.show()
```

```python
# classification_report
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
