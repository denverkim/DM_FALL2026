# 규칙기반 챗봇

```python
def chatbot(user_input):
    if "안녕" in user_input:
        return "안녕하세요! 무엇을 도와드릴까요?"

    elif "배송" in user_input:
        return "주문번호를 입력해 주세요."

    elif "환불" in user_input:
        return "환불 절차를 안내해 드리겠습니다."

    elif "영업시간" in user_input:
        return "영업시간은 오전 9시부터 오후 6시까지입니다."

    else:
        return "죄송합니다. 이해하지 못했습니다."

while True:
    user = input("사용자: ")

    if user == "종료":
        print("챗봇: 이용해 주셔서 감사합니다.")
        break

    print("챗봇:", chatbot(user))
```

# 글라디오로 챗봇 만들기

```python
import gradio as gr

# 규칙 기반 챗봇 함수
def chatbot(user_input, history):
    user_input = user_input.strip()

    if "안녕" in user_input:
        response = "안녕하세요! 무엇을 도와드릴까요?"

    elif "배송" in user_input:
        response = "주문번호를 입력해 주세요."

    elif "환불" in user_input:
        response = "환불 절차를 안내해 드리겠습니다."

    elif "영업시간" in user_input:
        response = "영업시간은 오전 9시부터 오후 6시까지입니다."

    elif "고마워" in user_input or "감사" in user_input:
        response = "도움이 되어 기쁩니다!"

    else:
        response = "죄송합니다. 이해하지 못했습니다."

    return response

# Gradio 인터페이스
demo = gr.ChatInterface(
    fn=chatbot,
    title="규칙 기반 챗봇",
    description="키워드 규칙에 따라 답변하는 간단한 챗봇입니다.",
    examples=[
        "안녕하세요",
        "배송 조회하고 싶어요",
        "환불하고 싶어요",
        "영업시간이 어떻게 되나요?",
        "감사합니다"
    ]
)

# 실행
if __name__ == "__main__":
    demo.launch()
```
