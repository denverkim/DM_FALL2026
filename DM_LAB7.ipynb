```python
# european_city_distances.csv 읽기
import pandas as pd

city = pd.read_csv('/content/european_city_distances.csv', delimiter=';', header=None)
city.head()
```

# 전처리

```python
# 인덱스 설정
city.set_index(0, inplace=True)
city.head()
```

```python
# 널값 제거
city.isna().sum()
city.dropna(axis=1, inplace=True) #열로 제거
city.head()
```

```python
# 컬럼이름 설정
city.columns = city.index
city.head()
```

```python
# PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
df = pca.fit_transform(city) #주성분
df = pd.DataFrame(df)
df.columns = ['PC1', 'PC2']
df.head()
```

```python
# 2D projection
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.scatter(df.PC1,df.PC2)
plt.grid()
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('European City Distances PCA')
for x, y, c in zip(df.PC1, df.PC2, city.index):
    plt.annotate(xy=[x,y], text=c, ha='center', xytext=[0,10], textcoords='offset points')
plt.ylim(-2000,3500)
plt.show()
```
